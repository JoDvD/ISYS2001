### Activity 1: Design an academic standing system

Your university needs a system that classifies students by their GPA. Before writing any code, think it through.

**Design questions:**

- What input does the system need?
- What output should it produce?
- Where are the decision points?

> **Your analysis** (double-click to edit):

```
Input needed: Student's GPA (a floating-point number)

Output needed: Academic standing status (a string, e.g., 'Dean's List', 'Good Standing', 'Academic Probation', 'Failing')

Decision points: The GPA thresholds that determine each academic standing category.
```

**Now define the logic.** How should a university classify students?

- What GPA ranges make sense?
- What happens at the boundaries (exactly 3.0, exactly 3.5)?
- What categories are actually useful for a student to hear?

> **Your classification rules** (double-click to edit):

```
If GPA is >= 3.5, then status is Dean's List
If GPA is >= 3.0 and < 3.5, then status is Good Standing
If GPA is >= 2.0 and < 3.0, then status is Academic Probation
If GPA is < 2.0, then status is Failing
```

**Test your logic before you code it.** Predict the output for each student:

### Working with AI

Now that you have done the thinking, this is where AI earns its keep. Notice the difference:

- **Useful:** "I have designed a GPA classifier with these ranges: [your ranges]. Help me implement this logic in Python using if/elif/else."
- **Less useful:** "Write a GPA calculator."

Try your version with your preferred AI tool, then compare its code to your logic. Did it match your thinking? What did it do differently?

````python
# Implement your academic standing system here

try:
    gpa = float(input("Enter student's GPA (0.0-4.0): "))

    if gpa >= 3.5:
        status = "Dean's List"
    elif gpa >= 3.0: # This implies gpa is between 3.0 and 3.5 (exclusive of 3.5)
        status = "Good Standing"
    elif gpa >= 2.0: # This implies gpa is between 2.0 and 3.0 (exclusive of 3.0)
        status = "Academic Probation"
    else:
        status = "Failing"

    print(f"For GPA {gpa}, the academic standing is: {status}")
except ValueError:
    print("Invalid input. Please enter a numerical GPA.")
````

### Activity 2: Design a study recommendation system

A learning app wants to give students personalised study advice. You design the decision logic.

Pick 2 to 3 factors that matter most to you, for example:

- Hours studied today
- Stress level
- Days until the next exam
- Hours of sleep last night

> **Your design** (double-click to edit):

```

- Hours studied today (integer)
- Days until the next exam (integer)

My rules:

IF Days until next exam is < 3 AND Hours studied today is < 2 THEN recommend 'Intense Study Session: Focus on key topics!'

ELIF Hours studied today is >= 3 THEN recommend 'Well Done! Consider a short break or light review.'
```

### Activity 3: Debug the broken classifier

Here is some code that looks fine but misbehaves. Do not fix it by guessing. Trace the logic first.

Run the cell below. It should give "Good Standing" for a GPA of 3.2. Does it?

````python
student_gpa = 3.2

if student_gpa >= 2.0:
    status = "Probation"
elif student_gpa >= 3.0:
    status = "Good Standing"
elif student_gpa >= 3.5:
    status = "Dean's List"
else:
    status = "Failing"

print(f"Status: {status}")
````

**Trace it, then fix it.**

- When `student_gpa = 3.2`, is `3.2 >= 2.0` true? What happens the moment a condition is true?
- Why does the order of the conditions matter here?

> **Your analysis** (double-click to edit):

```
The problem is: For a GPA of 3.2, the current code outputs "Probation" instead of the expected "Good Standing".

It happens because: The `if/elif` structure checks conditions sequentially. The first condition `student_gpa >= 2.0` is true for 3.2, so it assigns "Probation" and the rest of the `elif` conditions are never checked. The conditions are not ordered from most specific to least specific.

```

Now write your corrected version below.

````python
# Your fixed version here

student_gpa = 3.2

if student_gpa >= 3.5:
    status = "Dean's List"
elif student_gpa >= 3.0:
    status = "Good Standing"
elif student_gpa >= 2.0:
    status = "Probation"
else:
    status = "Failing"

print(f"Status: {status}")
````

**A better way to ask AI for debugging help.**

### Activity 5: Build an admissions decision system

Universities weigh several factors to make admission decisions. Design and build a fair, logical version.

**Design first:**

1. Choose 3 to 4 factors (GPA, test scores, activities, essay, special circumstances)
2. Decide how they combine (all required? at least 2 of 3?)
3. Set specific thresholds
4. Think about edge cases and tie-breakers

> **Your criteria** (double-click to edit):

```
Factor 1: GPA (float) - Must be 3.0 or higher.
Factor 2: Test Score (integer) - Must be 1200 or higher.
Factor 3: Extracurricular Activities (boolean) - Must have participated (True).
Factor 4: Essay Score (integer) - Must be 8 or higher (out of 10).

Decision logic:
Admitted if (GPA >= 3.0 AND Test Score >= 1200) AND (Extracurricular Activities is True OR Essay Score >= 8).

This means: High academic standing (GPA and Test Score) is mandatory, AND either strong extracurriculars OR an excellent essay.
```

````python
# Your admissions decision system

try:
    gpa = float(input("Enter applicant's GPA (0.0-4.0): "))
    test_score = int(input("Enter applicant's Test Score (e.g., SAT/ACT combined score): "))
    extracurriculars = input("Has participated in extracurricular activities? (yes/no): ").lower() == 'yes'
    essay_score = int(input("Enter applicant's Essay Score (1-10): "))

    # Evaluate conditions
    academic_standing_met = (gpa >= 3.0 and test_score >= 1200)
    other_factors_met = (extracurriculars or essay_score >= 8)

    if academic_standing_met and other_factors_met:
        decision = "Admitted"
    else:
        decision = "Not Admitted"

    print(f"Admission Decision: {decision}")

except ValueError:
    print("Invalid input. Please enter appropriate numerical or 'yes/no' values.")

# Test cases:
# - GPA 3.5, Test 1300, EC yes, Essay 9 -> Admitted
# - GPA 3.1, Test 1100, EC yes, Essay 8 -> Not Admitted (Test Score too low)
# - GPA 2.9, Test 1250, EC yes, Essay 9 -> Not Admitted (GPA too low)
# - GPA 3.2, Test 1200, EC no, Essay 7 -> Not Admitted (Neither EC nor Essay_score meet criteria)
# - GPA 3.2, Test 1200, EC yes, Essay 7 -> Admitted
# - GPA 3.2, Test 1200, EC no, Essay 8 -> Admitted
````

### Activity 6: Debug the tricky one

A student wrote this scholarship checker. The rule they wanted:

> Eligible if GPA is 3.0 or higher, AND (they have 50+ community service hours OR financial need), AND the essay scored 7 or higher.

Run it and test it. It does not behave the way they intended.

````python
print("=== Scholarship Eligibility Checker ===")

gpa = float(input("Enter GPA (0.0-4.0): "))
community_hours = int(input("Community service hours: "))
financial_need = input("Have financial need? (yes/no): ").lower()
essay_score = int(input("Essay score (1-10): "))

# Intended rule:
# GPA 3.0+ AND (community hours 50+ OR financial need) AND essay 7+

if gpa >= 3.0 and community_hours >= 50 or financial_need == "yes" and essay_score >= 7:
    print("Eligible for scholarship!")
else:
    print("Not eligible")
````

**Work through it.**

- Write the intended rule in plain English (you have it above).
- Python evaluates `and` before `or`. So how does the computer actually group that condition? Add brackets to show the grouping Python is really using.
- Where does that grouping differ from what the student meant?

> **Your analysis** (double-click to edit):

```
Intended rule: (gpa >= 3.0) AND (community_hours >= 50 OR financial_need == "yes") AND (essay_score >= 7)

What Python actually checks due to operator precedence (and before or):
(gpa >= 3.0 AND community_hours >= 50) OR (financial_need == "yes" AND essay_score >= 7)

Why that is wrong: The student intended for GPA and essay score to always be required, and for community hours OR financial need to be the middle condition. Python's default precedence groups 'and' operations before 'or' operations, which changes the logical structure. For example, a student with low GPA but high essay score and financial need could be incorrectly deemed eligible.

The fix: Use parentheses to explicitly group the conditions as intended, overriding default operator precedence.
```

Write your corrected version below.

````python
# Your corrected scholarship checker here

print("=== Scholarship Eligibility Checker ===")

try:
    gpa = float(input("Enter GPA (0.0-4.0): "))
    community_hours = int(input("Community service hours: "))
    financial_need = input("Have financial need? (yes/no): ").lower()
    essay_score = int(input("Essay score (1-10): "))

    # Intended rule:
    # GPA 3.0+ AND (community hours 50+ OR financial need) AND essay 7+

    if (gpa >= 3.0) and (community_hours >= 50 or financial_need == "yes") and (essay_score >= 7):
        print("Eligible for scholarship!")
    else:
        print("Not eligible")

except ValueError:
    print("Invalid input. Please ensure GPA, community hours, and essay score are numbers, and financial need is 'yes' or 'no'.")


# Test Cases for the corrected version:
# - Eligible: GPA 3.5, Comm. Hours 60, Fin. Need no, Essay 8 -> Eligible
# - Eligible: GPA 3.2, Comm. Hours 0, Fin. Need yes, Essay 7 -> Eligible
# - Not Eligible (GPA too low): GPA 2.9, Comm. Hours 60, Fin. Need no, Essay 8 -> Not Eligible
# - Not Eligible (Essay too low): GPA 3.5, Comm. Hours 60, Fin. Need no, Essay 6 -> Not Eligible
# - Not Eligible (Neither Comm. Hours nor Fin. Need): GPA 3.5, Comm. Hours 40, Fin. Need no, Essay 8 -> Not Eligible
````

**Asking AI well here:**

> "This should check (A AND (B OR C) AND D) but it is not working. Can you explain how operator precedence is grouping my condition, and how to fix it?"

Follow up with: "Can you show me the same logic written with explicit brackets?"

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

**Working with AI**
- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
- Give it your intent when you ask for help, especially when debugging.
