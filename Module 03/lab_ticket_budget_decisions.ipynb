## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Your task is to create a Python program that acts as a 'Purchase Advisor' tool. The program should guide a user through evaluating the affordability of a desired item based on their financial situation.

Here are the specific requirements for the program:

1.  **Input Collection:**
    *   The program must first greet the user with a welcoming message.
    *   It should then clearly prompt the user to enter their `monthly_income` (as a numerical value, e.g., '2500').
    *   Next, it should prompt the user to enter their `current_savings` (as a numerical value, e.g., '10000').
    *   Finally, it should prompt the user to enter the `item_cost` of the desired purchase (as a numerical value, e.g., '300').
    *   All inputs should be converted to appropriate numeric types (e.g., float or int) for calculations.

2.  **Affordability Logic and Decision Making:**
    *   The program must implement a series of `if-elif-else` statements to evaluate affordability based on the collected inputs. The decision rules are as follows:
        *   **Rule 1: Easily Affordable**
            *   If `item_cost` is less than or equal to 10% of `monthly_income` (i.e., `item_cost <= 0.10 * monthly_income`).
            *   **Recommendation:** Print a message like: "Great news! This purchase is easily affordable and won't significantly impact your finances."

        *   **Rule 2: Affordable with Planning**
            *   If `item_cost` is greater than 10% but less than or equal to 50% of `monthly_income` (i.e., `0.10 * monthly_income < item_cost <= 0.50 * monthly_income`).
            *   AND `current_savings` are at least 3 times the `monthly_income` (i.e., `current_savings >= 3 * monthly_income`).
            *   **Recommendation:** Print a message like: "This purchase is affordable, but it's a notable expense. Consider if it aligns with your budget and financial goals."

        *   **Rule 3: Possible, but Save More**
            *   If `item_cost` is greater than 10% but less than or equal to 50% of `monthly_income` (same as Rule 2).
            *   AND `current_savings` are less than 3 times the `monthly_income` (i.e., `current_savings < 3 * monthly_income`).
            *   **Recommendation:** Print a message like: "This purchase is possible, but it would be wise to increase your savings buffer before proceeding. Aim for at least 3 months of income in savings."

        *   **Rule 4: Potentially Too Expensive (Lower Savings)**
            *   If `item_cost` is greater than 50% of `monthly_income` (i.e., `item_cost > 0.50 * monthly_income`).
            *   AND `current_savings` are less than 5 times the `monthly_income` (i.e., `current_savings < 5 * monthly_income`).
            *   **Recommendation:** Print a message like: "This is a significant expense and might be too much for your current financial situation. It's recommended to reconsider or save substantially more."

        *   **Rule 5: Large Purchase, but Covered (Higher Savings)**
            *   If `item_cost` is greater than 50% of `monthly_income` (same as Rule 4).
            *   AND `current_savings` are at least 5 times the `monthly_income` (i.e., `current_savings >= 5 * monthly_income`).
            *   **Recommendation:** Print a message like: "This is a large purchase. While you have ample savings to cover it, ensure it's a well-thought-out investment or expense."

3.  **Output:**
    *   The program should always conclude with a summary of the inputs and the final recommendation, clearly formatted.
    *   Example output format: "Your monthly income: $X.XX, Current savings: $Y.YY, Item cost: $Z.ZZ. Recommendation: [Decision message here]"

4.  **Error Handling (Basic):**
    *   Implement basic error handling for non-numeric inputs using a `try-except` block for each input. If a non-numeric value is entered, print an error message and exit or prompt again (exiting is fine for this exercise).

5.  **Code Structure:**
    *   Encapsulate the entire logic within a single Python function, e.g., `run_purchase_advisor()`, and then call this function to execute the program.

Ensure the code is clean, readable, and directly implements these rules without additional assumptions.

````python
def run_purchase_advisor():
    print("Welcome to the Purchase Advisor! Let's evaluate your potential purchase.")

    # 1. Input Collection
    try:
        monthly_income_str = input("Please enter your monthly income (e.g., 2500): ")
        monthly_income = float(monthly_income_str)

        current_savings_str = input("Please enter your current savings (e.g., 10000): ")
        current_savings = float(current_savings_str)

        item_cost_str = input("Please enter the item cost of your desired purchase (e.g., 300): ")
        item_cost = float(item_cost_str)
    except ValueError:
        print("Error: Invalid input. Please enter numeric values for income, savings, and item cost. Exiting.")
        return

    # 2. Affordability Logic and Decision Making
    recommendation = ""

    # Rule 1: Easily Affordable
    if item_cost <= 0.10 * monthly_income:
        recommendation = "Great news! This purchase is easily affordable and won't significantly impact your finances."
    # Rule 2: Affordable with Planning
    elif 0.10 * monthly_income < item_cost <= 0.50 * monthly_income and current_savings >= 3 * monthly_income:
        recommendation = "This purchase is affordable, but it's a notable expense. Consider if it aligns with your budget and financial goals."
    # Rule 3: Possible, but Save More
    elif 0.10 * monthly_income < item_cost <= 0.50 * monthly_income and current_savings < 3 * monthly_income:
        recommendation = "This purchase is possible, but it would be wise to increase your savings buffer before proceeding. Aim for at least 3 months of income in savings."
    # Rule 4: Potentially Too Expensive (Lower Savings)
    elif item_cost > 0.50 * monthly_income and current_savings < 5 * monthly_income:
        recommendation = "This is a significant expense and might be too much for your current financial situation. It's recommended to reconsider or save substantially more."
    # Rule 5: Large Purchase, but Covered (Higher Savings)
    elif item_cost > 0.50 * monthly_income and current_savings >= 5 * monthly_income:
        recommendation = "This is a large purchase. While you have ample savings to cover it, ensure it's a well-thought-out investment or expense."

    # 3. Output
    print(f"\nYour monthly income: ${monthly_income:.2f}, Current savings: ${current_savings:.2f}, Item cost: ${item_cost:.2f}. Recommendation: {recommendation}")

# Call the function to execute the program
run_purchase_advisor()
````
