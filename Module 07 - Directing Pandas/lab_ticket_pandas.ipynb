## Task 1: total sales by region, highest first

Load the file into a DataFrame, clean whatever needs cleaning so the sale values are
real numbers, and produce the total `Total_Sale` for each `Region`, ordered from
highest to lowest.

Plan it in plain words in the cell below before you write the pandas. If you use an
assistant to help with a line, give it your plan and make sure you can explain what
it hands back.

My plan, in plain words:

*First, I will load the sales_data.csv file into a pandas DataFrame. Next, I will clean the Total_Sale column by removing the dollar sign and converting it to a numeric data type. After that, I will group the DataFrame by the Region column and calculate the sum of Total_Sale for each region. Finally, I will sort the results in descending order to show the regions with the highest total sales first, and then display the result.*


```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/sales_data.csv')

# Clean 'Total_Sale' column: remove '$' and convert to float
df['Total_Sale'] = df['Total_Sale'].replace('[$,]', '', regex=True).astype(float)

# Clean 'Quantity' column: remove '$' and convert to integer
df['Quantity'] = df['Quantity'].replace('[$,]', '', regex=True).astype(int)

# Calculate total sales by region and sort in descending order
total_sales_by_region = df.groupby('Region')['Total_Sale'].sum().sort_values(ascending=False)

# Display the result
print("Total Sales by Region (highest first):")
display(total_sales_by_region)
```

## Task 2: units sold per product

Now find how many units of each product were sold in total, using the `Quantity`
column.

This is where looking at the data first pays off. If your numbers come out looking
like money rather than counts of units, stop and work out why before you go on. A
correct answer here should read as sensible unit counts, not dollar amounts.

My plan, in plain words:

*I will use the existing DataFrame. I need to calculate the total units sold for each unique Product. To do this, I will group the DataFrame by the Product column and then sum the Quantity for each group. The result will show how many units of each product were sold in total.*


```python
# Calculate total units sold per product
total_units_per_product = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)

# Display the result
print("Total Units Sold per Product:")
display(total_units_per_product)
```
