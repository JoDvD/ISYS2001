## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:
Write a Python program that functions as a simple budget tracker. The program should:
1. Prompt the user to enter their monthly income.
2. Prompt the user to enter up to three individual expenses. For each expense, ask for its name (e.g., 'Rent', 'Groceries') and its amount.
3. Calculate the total expenses.
4. Calculate the remaining balance by subtracting total expenses from the monthly income.
5. Print a clear summary that includes:
   - The monthly income.
   - Each expense with its name and amount.
   - The total expenses.
   - The remaining balance.
6. If the remaining balance is negative, display a warning message indicating that the user is over budget and by how much.
7. Ensure that all monetary inputs are handled as numeric values (e.g., floats).
"""
```

### Paste and Test Your AI-Generated Budget Tracker

```python
# AI-GENERATED BUDGET TRACKER CODE

def get_float_input(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Please enter a non-negative value.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a numeric value.")

def get_expense_details(expense_number):
    name = input(f"Enter name for expense {expense_number}: ")
    amount = get_float_input(f"Enter amount for {name}: $")
    return name, amount

# 1. Prompt the user to enter their monthly income.
monthly_income = get_float_input("Enter your monthly income: $")

expenses = []
total_expenses = 0.0

# 2. Prompt the user to enter up to three individual expenses.
for i in range(1, 4):
    add_expense = input(f"Do you want to add expense {i}? (yes/no): ").lower()
    if add_expense == 'yes':
        expense_name, expense_amount = get_expense_details(i)
        expenses.append({'name': expense_name, 'amount': expense_amount})
        total_expenses += expense_amount
    else:
        break # Exit loop if user doesn't want to add more expenses

# 4. Calculate the remaining balance.
remaining_balance = monthly_income - total_expenses

# 5. Print a clear summary.
print("\n--- Budget Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print("Expenses:")
if not expenses:
    print("  No expenses added.")
else:
    for expense in expenses:
        print(f"  - {expense['name']}: ${expense['amount']:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Balance: ${remaining_balance:.2f}")

# 6. If the remaining balance is negative, display a warning message.
if remaining_balance < 0:
    print(f"\n⚠️ Warning: You are over budget by ${abs(remaining_balance):.2f}!")
```

```python
# REFINED VERSION — your working budget tracker

def get_float_input(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Please enter a non-negative value.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a numeric value.")

def get_expense_details_refined(expense_number):
    while True:
        name = input(f"Enter name for expense {expense_number} (e.g., 'Rent', 'Groceries', or leave blank to finish): ")
        if not name.strip(): # Check if name is empty or just whitespace
            return None, None # Signal to stop adding expenses
        amount = get_float_input(f"Enter amount for {name}: $")
        return name, amount

# 1. Prompt the user to enter their monthly income.
monthly_income = get_float_input("Enter your monthly income: $")

expenses = []
total_expenses = 0.0

# 2. Prompt the user to enter up to three individual expenses.
for i in range(1, 4):
    expense_name, expense_amount = get_expense_details_refined(i)
    if expense_name is None: # User chose to stop adding expenses
        break
    expenses.append({'name': expense_name, 'amount': expense_amount})
    total_expenses += expense_amount

# 4. Calculate the remaining balance.
remaining_balance = monthly_income - total_expenses

# 5. Print a clear summary.
print("\n--- Budget Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print("Expenses:")
if not expenses:
    print("  No expenses added.")
else:
    for expense in expenses:
        print(f"  - {expense['name']}: ${expense['amount']:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Balance: ${remaining_balance:.2f}")

# 6. If the remaining balance is negative, display a warning message.
if remaining_balance < 0:
    print(f"\n⚠️ Warning: You are over budget by ${abs(remaining_balance):.2f}!")
```

## Part 4: Enhancement Challenge

Pick ONE feature to add with AI's help:

```python
# Document your enhancement choice and the prompt you used
enhancement_choice = "I chose to add: Percentage of income spent"
enhancement_prompt = """
Enhance the existing Python budget tracker program. After calculating the total expenses and remaining balance, calculate the percentage of the monthly income that is spent on expenses. Display this percentage in the budget summary. Handle cases where monthly income might be zero to avoid division by zero errors.
"""

print(f"🚀 Enhancement: {enhancement_choice}")
```

```python
# ENHANCED BUDGET TRACKER

def get_float_input(prompt):
    while True:
        try:
            value = float(input(prompt))
            if value < 0:
                print("Please enter a non-negative value.")
                continue
            return value
        except ValueError:
            print("Invalid input. Please enter a numeric value.")

def get_expense_details_refined(expense_number):
    while True:
        name = input(f"Enter name for expense {expense_number} (e.g., 'Rent', 'Groceries', or leave blank to finish): ")
        if not name.strip(): # Check if name is empty or just whitespace
            return None, None # Signal to stop adding expenses
        amount = get_float_input(f"Enter amount for {name}: $")
        return name, amount

# 1. Prompt the user to enter their monthly income.
monthly_income = get_float_input("Enter your monthly income: $")

expenses = []
total_expenses = 0.0

# 2. Prompt the user to enter up to three individual expenses.
for i in range(1, 4):
    expense_name, expense_amount = get_expense_details_refined(i)
    if expense_name is None: # User chose to stop adding expenses
        break
    expenses.append({'name': expense_name, 'amount': expense_amount})
    total_expenses += expense_amount

# 4. Calculate the remaining balance.
remaining_balance = monthly_income - total_expenses

# New: Calculate percentage of income spent
percentage_spent = 0.0
if monthly_income > 0:
    percentage_spent = (total_expenses / monthly_income) * 100

# 5. Print a clear summary.
print("\n--- Budget Summary ---")
print(f"Monthly Income: ${monthly_income:.2f}")
print("Expenses:")
if not expenses:
    print("  No expenses added.")
else:
    for expense in expenses:
        print(f"  - {expense['name']}: ${expense['amount']:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Balance: ${remaining_balance:.2f}")
print(f"Percentage of income spent: {percentage_spent:.2f}%")

# 6. If the remaining balance is negative, display a warning message.
if remaining_balance < 0:
    print(f"\n⚠️ Warning: You are over budget by ${abs(remaining_balance):.2f}!")
```
