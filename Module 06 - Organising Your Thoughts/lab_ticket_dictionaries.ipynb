## The three functions your code has to get right

Build your dictionary however you like, then write these three functions to work on it.
The checks at the bottom call them by name, so use these names and have them behave as
described:

- `total_for(spending, category)` returns the amount spent in that category, and returns
  `0` for a category that was never recorded. It must not crash on a missing category.
- `add_transaction(spending, transaction)` takes one transaction, in the same shape as
  the records in the list above, and updates your dictionary so that amount is counted
  from then on.
- `summary(spending)` returns a dictionary that maps every category to its total.

How `spending` is built and shaped inside is entirely your decision. The checks only care
about what these three functions do, not how you store things underneath.

```python
# Build your dictionary from the transactions, and write your three functions here.
spending = {}

# Populate the spending dictionary from the initial transactions list
for t in transactions:
    category = t['category']
    amount = t['amount']
    if category not in spending:
        spending[category] = []
    spending[category].append(amount)


def total_for(spending_data, category):
    """Returns the total amount spent for a given category."""
    return sum(spending_data.get(category, []))

def add_transaction(spending_data, transaction):
    """Adds a new transaction to the spending dictionary."""
    category = transaction['category']
    amount = transaction['amount']
    spending_data.setdefault(category, []).append(amount)

def summary(spending_data):
    """Returns a dictionary mapping each category to its total spending."""
    report = {}
    for category, amounts in spending_data.items():
        report[category] = sum(amounts)
    return report


print("Spending dictionary initialized and functions defined.")
# Display the initial summary to see the structure
print("Initial Spending Summary:", summary(spending))
```

```python
# Your own feature goes here.
# This feature helps understand where the most money is being allocated.

def find_highest_spending_category(spending_data):
    """Identifies the category with the highest total spending."""
    current_summary = summary(spending_data)
    if not current_summary:
        return None, 0

    highest_category = None
    max_spending = 0

    for category, total_amount in current_summary.items():
        if total_amount > max_spending:
            max_spending = total_amount
            highest_category = category

    return highest_category, max_spending

highest_cat, highest_amt = find_highest_spending_category(spending)

if highest_cat:
    print(f"The category with the highest spending is '{highest_cat}' with a total of ${highest_amt}.")
else:
    print("No spending data available to determine the highest spending category.")
```
